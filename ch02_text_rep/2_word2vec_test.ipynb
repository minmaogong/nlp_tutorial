```python
# 查询近义词
similar_words = model.similar_by_word('地铁', topn=5)
print(similar_words)
```

```python
similar_words = model.most_similar(positive=['地铁', '公交'], topn=5)
print(similar_words)
```

```python
# 对比填词
similar_words = model.most_similar(positive=['男人', '女孩'], negative=['女人'], topn=5)
print(similar_words)
```

## 自行训练词向量

```python
# 1. 准备语料数据
sentences = [['我', '每天','乘坐', '地铁', '上班'], ['我','每天', '乘坐', '公交', '上班']]
```

```python
# 2. 定义并训练模型
model = Word2Vec(
    sentences=sentences, # 已分词的句子序列
    vector_size=10, # 词向量维度
    window=2, # 上下文窗口大小
    min_count=1, # 最小词频（低于将被忽略）
    workers=4, # 并行训练线程数
    sg=1, # 1:Skip-Gram，0:CBOW
)
```

```python
print(model)
```

```python
print(model.wv)
```

```python
print(model.wv['地铁'])
```

```python
print(model.wv['公交'])
```

```python
print(model.wv.similarity('地铁', '公交'))
```

## 使用真实语料库训练词向量

```python
import pandas as pd
import jieba
```

```python
# 1. 加载数据
data = pd.read_csv('data/online_shopping_10_cats.csv').dropna()
data.head()
```

```python
# 2. 取评论数据，并分词
sentences = [ [token for token in jieba.lcut(sentence) if token.strip() != ''] for sentence in data['review'] ]
```

```python
print(sentences[:5])
```

```python
# 3. 定义模型并训练
model = Word2Vec(
    sentences=sentences, # 已分词的句子序列
    vector_size=100, # 词向量维度
    window=5, # 上下文窗口大小
    min_count=2, # 最小词频（低于将被忽略）
    workers=4, # 并行训练线程数
    sg=1 # 1:Skip-Gram，0:CBOW
)
```

```python
# 4. 查看词向量
print(model.wv['地铁'])
```

```python
print(model.wv.similarity('地铁', '公交'))
```

## 保存训练好的词向量到文件

```python
model_path = 'data/word2vec.kv'
model.wv.save_word2vec_format(model_path)
```

```python
kv_model = KeyedVectors.load_word2vec_format(model_path)
```
