```python
import torch
import torch.nn as nn
from gensim.models import KeyedVectors
```

## 词嵌入层

```python
embedding = nn.Embedding(num_embeddings=10000, embedding_dim=100)
```

```python
print(embedding.weight.shape)
```

## 加载已训练好的词向量

```python
# 1. 记载KeyedVectors模型
kv_model = KeyedVectors.load_word2vec_format('data/word2vec.kv')
```

```python
num_embeddings = len(kv_model.index_to_key)
print(num_embeddings)
embedding_dim = kv_model.vector_size
print(embedding_dim)
```

```python
# 2. 定义嵌入层，记载
# embedding = nn.Embedding(num_embeddings=num_embeddings, embedding_dim=embedding_dim)
embedding = nn.Embedding.from_pretrained(
    embeddings=torch.tensor(kv_model.vectors),
    freeze=False,
)
```

## 嵌入层测试

```python
# 1. 获取词表
word2id = kv_model.key_to_index
print(word2id)
```

```python
# 2. 获取输入，id
input_id = word2id['地铁']
print(input_id)
```

```python
input = torch.tensor(input_id)
print(input)
```

```python
# 3. 前向传播
vector = embedding(input)
print(vector)
```

## 测试示例

```python
import jieba
```

```python
# text = "我喜欢乘坐地铁"
text = "我喜欢乘坐宇宙飞船"
```

```python
# 1. 分词
tokens = jieba.lcut(text)
print(tokens)
```

## OOV问题（未登录词）

```python
# 在词表中增加处理OOV的特殊token
unk_token = '<UNK>'
id2word = [unk_token] + kv_model.index_to_key
print(id2word)
```

```python
# 转换字典
word2id = { word:id for id, word in enumerate(id2word) }
print(word2id)
```

```python
# 修改词嵌入层
embedding_matrix = torch.cat([torch.zeros(1, kv_model.vector_size), torch.tensor(kv_model.vectors)]) # 通过torch.cat将零向量和原始词向量拼接起来，形成新的嵌入矩阵
embedding = nn.Embedding.from_pretrained(
    embeddings=embedding_matrix,
    freeze=False,
)
print(embedding.weight)
```

```python
text = "我喜欢乘坐宇宙飞船"
```

```python
# 1. 分词
tokens = jieba.lcut(text)
print(tokens)
```

```python
# 2. 编码（id化）
ids = [word2id.get(token, word2id[unk_token]) for token in tokens]
print(ids)
```

```python
# 3. 定义模型输入
input = torch.tensor(ids)
print(input)
```

```python
# 4. 模型前向传播
vectors = embedding(input)
print(vectors)
print(vectors.shape)
```
