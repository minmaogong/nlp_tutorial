```python
import jieba
```

```python
text = "小明毕业于北京大学计算机系"
```

## 精确模式

```python
words_generator = jieba.cut(text) # 返回一个生成器
```

```python
for word in words_generator:
    print(word)
```

```python
# 直接返回分词后的列表
word_list = jieba.lcut(text)
print(word_list)
```

## 全模式

```python
words_generator = jieba.cut(text, cut_all=True)
for word in words_generator:
    print(word)
```

```python
word_list = jieba.lcut(text, cut_all=True)
print(word_list)
```

## 搜索引擎模式

```python
word_list = jieba.lcut_for_search(text)
print(word_list)
```

## 自定义词典

```python
text = "随着云计算技术的普及，越来越多企业开始采用云原生架构来部署服务，并借助大模型能力提升智能化水平，实现业务流程的自动化与智能决策。"
```

```python
word_list = jieba.lcut(text)
print(word_list)
```

```python
# 加载自定义词典
jieba.load_userdict('data/user_dict.txt')
```

```python
word_list = jieba.lcut(text)
print(word_list)
```

```python
# 添加一个新词
jieba.add_word("越来越多", freq=100)
```
