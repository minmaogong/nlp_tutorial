```python
from transformers import AutoModel, AutoModelForSequenceClassification
import torch
```

```python
model = AutoModel.from_pretrained("google-bert/bert-base-chinese")
```

```python
print(model)
```

```python
print(model.embeddings.word_embeddings.weight)
```

```python
print(model.embeddings.word_embeddings.weight.shape)
```

## 加载带任务头的模型

```python
model_task = AutoModelForSequenceClassification.from_pretrained("google-bert/bert-base-chinese", num_labels=3)
```

```python
print(model_task)
```

## 模型的输入输出

```python
vocab_size = 1000
batch_size = 32
seq_len = 12
# 定义输入(N, L)
input_ids = torch.randint(vocab_size, size = (batch_size, seq_len))
```

```python
# 前向传播
output = model(input_ids)
```

```python
print(output)
```

```python
print(output.last_hidden_state.shape) # last_hidden_state 是BERT中 BertEncoder层 的输出，形状为 (batch_size, seq_len, hidden_size)
```

```python
print(output.pooler_output.shape) # pooler_output 是BERT的 BertPooler层，获取classification token [CLS] 进行线形转换和非线性转换后的输出，形状为 (batch_size, hidden_size)
```

## 带任务头的模型输入输出

```python
output_task = model_task(input_ids)
```

```python
print(output_task)
```

```python
# 定义目标值
target = torch.randint(3, size=(batch_size,))
print(target)
```

```python
output_task = model_task(input_ids, labels=target)
print(output_task)
```

```python
print(output_task.logits.shape) # logits 是模型前向传播的输出，不是经过softmax的概率分布，形状为 (batch_size, num_labels)。后面通过CrossEntropyLoss计算损失时，会自动将logits输入到softmax中，softmax的输出才是概率分布。
```
