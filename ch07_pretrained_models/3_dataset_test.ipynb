## 数据预处理

```python
# 1. 删除列
dataset3 = dataset3.remove_columns(['cat'])
print(dataset3)
```

```python
# 2. 过滤行
dataset3 = dataset3.filter(lambda x: x['review'] is not None)
print(dataset3)
```

```python
# 3. 划分数据集
dataset_dict = dataset3.train_test_split(test_size=0.2)
print(dataset_dict)
```

```python
# 4. 对文本数据进行编码转换（id化）
# 4.1 定义数据集
train_dataset = dataset_dict['train']
print(train_dataset)
```

```python
print(train_dataset[:3])
```

```python
# 4.2 定义分词器
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("google-bert/bert-base-chinese")
```

```python
# 4.3 编码转换review
def batch_encode(batch):
    inputs = tokenizer(batch['review'], padding='max_length', max_length=128, truncation=True, return_tensors="pt") # padding=True 表示以样本数据batch['review']中长度最长的文本为基准进行padding，return_tensors="pt" 表示返回PyTorch张量
    inputs['labels'] = batch['label']
    return inputs

train_dataset = train_dataset.map(batch_encode, batched=True, batch_size=128, remove_columns=['label', 'review']) # batch_size 表示每次映射处理的样本数量，batched=True 表示将数据集划分为多个批次进行映射处理，batch_encode函数会对每个批次的数据进行编码转换
```

```python
print(train_dataset)
```

```python
print(train_dataset[:3])
```

```python
# 5. 转换为Tensor
train_dataset.set_format(type='torch')
print(train_dataset[:3])
```

```python
dataset_dict=dataset_dict.map(batch_encode, batched=True, batch_size=128, remove_columns=['label', 'review'])
print(dataset_dict)
```
