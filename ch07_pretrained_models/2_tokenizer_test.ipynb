```python
from transformers import AutoTokenizer, AutoModel, AutoModelForSequenceClassification
```

## 加载分词器

```python
tokenizer = AutoTokenizer.from_pretrained("google-bert/bert-base-chinese")
```

```python
print(type(tokenizer))
print(tokenizer.is_fast)
print(tokenizer)
```

## 使用分词器

```python
text = "我爱自然语言处理"
```

```python
# 1. 分词
tokens = tokenizer.tokenize(text)
print(tokens)
```

```python
# 2. id化(编码)
ids = tokenizer.convert_tokens_to_ids(tokens)
print(ids)
```

```python
# 3. 编码encode = 分词 + id化，还可以加入特殊token
ids1 = tokenizer.encode(text, add_special_tokens=False)
print(ids1)
```

```python
ids2 = tokenizer.encode(text)
print(ids2)
```

```python
# padding: True/longest: 取最长的句子长度进行填充 max_length: 取指定的最大长度进行填充 do_not_pad: 不进行填充
# max_length: 当需要填充时，如果没有设置max_length，会使用config文件中设置的max_length，如果设置了max_length，则使用max_length进行填充和截断
# truncation: True/False 超出max_length是否截断
ids3 = tokenizer.encode(text, padding="max_length", max_length=15, truncation=True)
print(ids3)
```

```python
ids3 = tokenizer.encode(text, padding="max_length", max_length=8, truncation=True)
print(ids3)
```

```python
ids4 = tokenizer.encode(text, padding="max_length", truncation=True, return_tensors="pt") # pt 就返回 pytorch tensor 张量，np 就返回 numpy ndarray
print(ids4)
```

```python
# 4. 反id化，id -> token
tokens = tokenizer.convert_ids_to_tokens(ids4.squeeze(0).tolist())
print(tokens)
```

```python
# 5. 解码 decode
# skip_special_tokens: True/False 是否跳过特殊token
str = tokenizer.decode(ids4.squeeze(0).tolist(), skip_special_tokens=True)
print(str)
```

```python
print(str.replace(" ", ""))
```

```python
# 6. 调用__call__方法，直接得到模型输入
inputs = tokenizer(text)
print(inputs)
```

```python
inputs = tokenizer(text, add_special_tokens=True, padding="max_length", max_length=15, truncation=True, return_tensors="pt")
print(inputs)
```

```python
text_pair = "我爱打篮球"
inputs = tokenizer(text=text, text_pair=text_pair, padding="max_length", max_length=20, return_tensors="pt")
print(inputs)
```

## 分词器与模型配合使用

```python
# 1. 准备数据
texts = ["我爱自然语言处理", "我爱打篮球", "你爱学习"]
```

```python
model_name = "google-bert/bert-base-chinese"
# 2. 加载模型和分词器
model= AutoModel.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
# 3. 利用分词器得到模型输入
inputs = tokenizer(texts, padding=True, return_tensors="pt")
print(inputs)
```

```python
# 4. 前向传播
outputs = model(input_ids=inputs["input_ids"], attention_mask=inputs["attention_mask"], token_type_ids=inputs["token_type_ids"])
print(outputs)
```

```python
outputs = model(**inputs) # **inputs 表示解包字典，将字典中的键值对作为关键字参数传递给函数
print(outputs)
```

## 带任务头的模型的配合使用

```python
# 1. 准备数据
texts = ["我爱自然语言处理", "我爱打篮球", "你爱学习"]
```

```python
model_name = "google-bert/bert-base-chinese"
# 2. 加载模型和分词器
model= AutoModelForSequenceClassification.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
# 3. 利用分词器得到模型输入
inputs = tokenizer(texts, padding=True, return_tensors="pt")
print(inputs)
```

```python
import torch

# 4. 在输入中添加labels
targets = [1, 0 ,1]
inputs["labels"] = torch.tensor(targets)
print(inputs)
```

```python
# 5. 前向传播
outputs = model(**inputs) # **inputs 表示解包字典，将字典中的键值对作为关键字参数传递给函数
print(outputs)
```

```python
# 6. 反向传播
outputs.loss.backward()
```
