```python
import math

import torch
```

```python
torch.set_printoptions(sci_mode=False, precision=6)
torch.manual_seed(42)
```

```python
SEQ_LEN = 4
embedding_dim = 4
dk = 128
```

```python
# 定义数据，形状(L, dk) dk -> 词向量维度
X = torch.randn(size=(SEQ_LEN, embedding_dim))
```

```python
# 计算Q、K、V矩阵
Wq = torch.randn(size=(embedding_dim, dk), requires_grad=True)
Wk = torch.randn(size=(embedding_dim, dk), requires_grad=True)
Wv = torch.randn(size=(embedding_dim, dk), requires_grad=True)
Q = X @ Wq
K = X @ Wk
V = X @ Wv
print(Q)
print(Q.shape)
```

```python
# 计算注意力评分
scores = Q @ K.T / math.sqrt(dk)
# scores = Q @ K.T
print(scores)
```

```python
# 计算注意力权重
weights = torch.softmax(scores, dim=-1)
print(weights)
```

```python
# 计算上下文向量
vectors = weights @ V
print(vectors)
print(vectors.shape)
```

```python
# 模拟损失计算，并反向传播
loss = vectors.sum()
loss.backward()
```
