```python
import torch
import torch.nn as nn
```

## 构造参数，创建Transformer

```python
model = nn.Transformer(
    d_model=64,
    nhead=8, # nhead要能被d_model整除。计算注意力评分的Q、K、V向量的维度为d_model/nhead=64/8=8。每一个head的输出最终会拼接起来
    num_encoder_layers=3,
    num_decoder_layers=3,
    batch_first=True,
)
```

## 前向传播

```python
d_model = 64
batch_size = 32
src_len = 12
tgt_len = 15
```

```python
# 定义数据：源序列和目标序列
src = torch.rand(size=(batch_size, src_len, d_model))
tgt = torch.rand(size=(batch_size, tgt_len, d_model))
```

```python
output = model(src, tgt)
print(output.shape) # 输出的形状与解码器的输入形状相同，为(batch_size, tgt_len, d_model)
```
