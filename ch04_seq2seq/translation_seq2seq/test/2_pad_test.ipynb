```python
import torch
from torch.nn.utils.rnn import pad_sequence
```

```python
vocab_size = 1000
# 定义3个长度不同的id序列
seq1 = torch.randint(vocab_size, (10, ))
seq2 = torch.randint(vocab_size, (5, ))
seq3 = torch.randint(vocab_size, (13, ))
print(seq1)
print(seq2)
print(seq3)
```

```python
# 构建一批数据的列表
seq_list = [seq1, seq2, seq3]
print(seq_list)
```

```python
# 填充，并合并成一个tensor
batch = pad_sequence(seq_list, batch_first=True, padding_value=0)
print(batch)
print(batch.shape)
```
