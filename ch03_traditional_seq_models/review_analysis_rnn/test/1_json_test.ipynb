```python
import pandas as pd
```

```python
# 构建一个DataFrame， 3列
df = pd.DataFrame(
    data=dict(A=range(1, 6), B=range(2, 7), C=range(5, 10)),
    columns=['A', 'B', 'C'],
)
print(df)
```

```python
# 保存成json文件
df.to_json('df.json', orient='records')
```

```python
df.to_json('df.jsonl', orient='records', lines=True)
```

```python
# 读取文件
data = pd.read_json('df.jsonl', lines=True)
print(data)
```

```python
print(data.to_dict(orient='records'))
```
