```python
import torch
import torch.nn as nn
```

## 定义RNN层

```python
# 1. 定义单层单向RNN层（模型）
rnn = nn.RNN(
    input_size=3,
    hidden_size=4,
    batch_first=True,
)
```

```python
# 1. 定义两层双向RNN层（模型）
rnn2 = nn.RNN(
    input_size=3,
    hidden_size=4,
    batch_first=True,
    num_layers=2,
    bidirectional=True,
)
```

## 前向传播

```python
# 1. 定义输入: (N, L, input_size)
input = torch.randn(2, 4, 3) # 以“文武双全”“玉树临风” 两条数据为例，每个句子有4个词（序列长度为4），每个词用3维向量表示
```

```python
# 2. 前向传播
output, hn = rnn(input)
```

```python
print(output.shape) # [2, 4, 4] 2表示batch_size，4表示序列长度，4表示hidden_size
print(hn.shape) # [1, 2, 4] 1表示单层，2表示batch_size，4表示hidden_size
```

```python
print(output)
```

```python
print(hn)
```

```python
# 两层双向RNN前向传播
output, hn = rnn2(input)
print(output.shape) #[2, 4, 8] 2表示batch_size，4表示序列长度，8表示hidden_size*2（双向）
print(hn.shape) #[4, 2, 4] 4表示num_layers*2（双向），2表示batch_size，4表示hidden_size
```

```python
print(output)
```

```python
print(hn)
```
