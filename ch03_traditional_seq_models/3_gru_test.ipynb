```python
import torch
import torch.nn as nn
```

## GRU 门控循环单元网络

```python
# 1. 定义输入：(N, L, hidden_size)
input = torch.randn(2, 4, 3) # 以“文武双全”“玉树临风” 两条数据为例，每个句子有4个词（序列长度为4），每个词用3维向量表示
```

```python
# 2. 定义两层双向GRU层（模型）
gru = nn.GRU(
    input_size=3, # 输入特征维度为3
    hidden_size=4, # 隐藏状态的大小为4
    num_layers=2, # GRU层数为2
    bidirectional=True, # 双向GRU
    batch_first=True # 输入数据的第一个维度是batch_size
)
```

```python
# 3. 前向传播
output, hn = gru(input)
```

```python
print(output.shape) # [2, 4, 8] 2表示batch_size，4表示序列长度，8表示hidden_size*2（双向）
print(hn.shape) # [4, 2, 4] 4表示num_layers*2（双向），2表示batch_size，4表示hidden_size
```
