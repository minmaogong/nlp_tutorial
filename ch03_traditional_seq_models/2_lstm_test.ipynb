```python
import torch
import torch.nn as nn
```

```python
# 1. 定义输入：(N, L, input_size)
input = torch.randn(2, 4, 3) # 以“文武双全”“玉树临风” 两条数据为例，每个句子有4个词（序列长度为4），每个词用3维向量表示
```

```python
# 2. 定义两层双向LSTM层（模型）
lstm = nn.LSTM(
    input_size=3,
    hidden_size=5, # 这里设置了隐藏状态的大小为5
    batch_first=True,
    num_layers=2,
    bidirectional=True,
    proj_size=4 # 这里设置了投影层的大小为4，表示输出的隐藏状态维度为4，而不是hidden_size*2=10
)
```

```python
# 3. 前向传播
output, (hn, cn) = lstm(input)
```

```python
print(output.shape) # [2, 4, 8] 2表示batch_size，4表示序列长度，8表示proj_size*2（双向）
print(hn.shape) # [4, 2, 4] 4表示num_layers*2（双向），2表示batch_size，4表示proj_size
print(cn.shape) # [4, 2, 5] 4表示num_layers*2（双向），2表示batch_size，5表示hidden_size
```
